```python
x = 5
y = 5.1

# NOTE: when using a fraction or a variable that holds one to exponenet by, it better to put it
# inside `{}` or it could cause some format issues, for example the `y` here without
# putting it inside `{}` and make something exponenet with it, it will put the `5` only
# and the `.1` will be inlined

display(Math("%g^{\\frac{3}{4}} \\times 4^{%g} = %g" %(x, (y), (x**(3/4) * 4**y) )))
display(Math("\\frac{3^3}{%g^{%g}} = %g" %(x, y, (3**3 / x**y) )))
display(Math("10^{%g-4} = %g" %(x, (10**(x-4)) ) ))
```

```python
base = 32

for i in range(0, 10):
    display(Math("\\text{%g to the power of %g is %g}" %(base, i, base**i)))
    print("%g to the power of %g is %g" %(base, i, base**i))
```
