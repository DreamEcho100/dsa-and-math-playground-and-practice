```python
mat = np.zeros([4, 6])

mat[0, 1] = 2
mat[2, 4] = 7

mat
```

```python
num_range = range(0,4)

for rowi in num_range:
    for coli in num_range:
        mat[rowi, coli] = (-1) ** (rowi+coli)

mat
```

```python
from sympy.abc import x, y
xy_set = range(0, 3)
xy_row_size = len(xy_set)
xy_col_size = len(xy_set)
mat = np.zeros([xy_row_size, xy_col_size])
expr = (4+x)*(2-y)


for x_num in xy_set: 
    for y_num in xy_set:
        result = expr.subs({ x: x_num, y: y_num })
        mat[x_num, y_num] = result
        # display(Math(
        #     "\\text{When x=%g and y=%g, f(x,y)=%g}" 
        #     %(x_num, y_num, result)
        # ))

display(Math(
    sym.latex(
        sym.sympify(mat)
    )
))
```
