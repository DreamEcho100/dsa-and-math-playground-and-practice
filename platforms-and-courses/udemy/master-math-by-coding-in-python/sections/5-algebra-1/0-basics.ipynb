```python
import sympy as sym
from IPython.display import display, Math
```

```python
x = sym.symbols("x")

expr = 2*x +4 -9
sym.solve(expr)
```

```python
display(Math(
'\\text{The soution to } %s \\text{ is  x = }%g'
%(
sym.latex(expr),
sym.solve(expr)[0]
)
))
```

```python
expr = x**2 - 4
sol = sym.solve(expr)

for i in range(0, len(sol)):
    display(Math(
        "\\text{Solution #%g is %s}"
        %(
            i + 1,
            sol[i]
        )
    ))
```

```python
y = sym.symbols("y")

expr = x / 4 - x*y + 5
display(Math(sym.latex(sym.solve(expr, y)[0])))
display(Math(sym.latex(sym.solve(expr, x)[0])))
```

```python
q = sym.symbols("q")
```

```python
expr = 3*q + 4/q + 3 - (5*q + 1/q + 1)
display(Math(
sym.latex(expr)
))
display(Math(
"q= %s" %(str(sym.latex(sym.solve(expr, q))))
))
```

```python
expr = 2*q +3*q**2 - 5/q - 4/q**3

display(Math(
sym.latex(expr)
))

display(Math(
sym.latex(expr.simplify())
))
display(Math(
sym.latex(expr.cancel())
))
```

```python
expr = (sym.sqrt(3) + sym.sqrt(15)*q) / (sym.sqrt(2) + sym.sqrt(10)*q)


display(Math(
sym.latex(expr)
))

expr.subs(q, 3).evalf()
display(Math(
sym.latex(expr.simplify())
))
```

```python
a, b, c, d = sym.symbols("a,b,c, d")
expr = (a+b)*(c+d)
display(Math(
sym.latex(expr.expand())
))
```

```python
from sympy.abc import x

term1 = (4*x + 5)
term2 = x
term3 = x - 7


display(Math(
sym.latex(term1*term3)
))
display(Math(
sym.latex((term1*term3).expand())
))
```

```python
from sympy.abc import x, y
expr = (4+x)*(2-y)

for x_num in range(0, 3): 
    for y_num in range(0, 3):
        result = expr.subs({ x: x_num, y: y_num })
        display(Math(
            "\\text{When x=%g and y=%g, f(x,y)=%g}" 
            %(x_num, y_num, result)
        ))
```
