```python
x, y = sym.symbols("x,y")
expr = x**2 + 4
for num in range(-2, 3):
    display(Math(
        "\\text{With x = %g" %(num) + ": }" +
        sym.latex(sym.sympify(expr)) +
        " \\quad \\Rightarrow \\quad " +
        # ( "\\," if num >= 0 else "") +
        "%g^2 + 4 = " %(num) +
        str(expr.subs(x, num))
    ))
```

```python
x, y, z = sym.symbols("x,y,z")
```

```python
ex1 = x**y * x**z
ex2 = x**y / x**z
ex3 = x**y * y**z

def tmp_format(ex):
    return sym.latex(ex), sym.latex(sym.simplify(ex))

display(Math( "%s = %s" %(tmp_format(ex1)) ))
display(Math( "%s = %s" %(tmp_format(ex2)) ))
display(Math( "%s = %s" %(tmp_format(ex3)) ))
```

```python
sym.Eq(sym.simplify(ex1 - sym.simplify(ex1)), 0)
```

```python
sym.Eq(sym.powsimp(ex1 - sym.simplify(ex1)), 0)
```
